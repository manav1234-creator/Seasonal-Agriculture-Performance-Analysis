```python
final_summary = df.groupby("Season").agg({
    "Rainfall_mm": "mean",
    "Avg_Temperature_C": "mean",
    "Soil_Moisture_pct": "mean",
    "Fertilizer_kg_ha": "mean",
    "Yield_Tonnes_Ha": "mean",
    "Production_Tonnes": "mean",
    "Water_Used_m3": "mean",
    "Water_Efficiency_t_per_1000m3": "mean",
    "Revenue_INR": "mean",
    "Total_Cost_INR": "mean",
    "Profit_INR": "mean",
    "Disease_Pest_Risk_pct": "mean"
})

final_summary
```

2. Seasonal agricultural planning should consider rainfall, temperature and soil moisture conditions.

3. Economic performance should be evaluated using profit rather than revenue alone.

4. Crop selection should consider seasonal productivity and environmental conditions.
